```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)

DATA_DIR = Path("../data_raw")
```

```python
plant1_gen = pd.read_csv(
    DATA_DIR / "Plant_1_Generation_Data.csv"
)

plant1_weather = pd.read_csv(
    DATA_DIR / "Plant_1_Weather_Sensor_Data.csv"
)

plant2_gen = pd.read_csv(
    DATA_DIR / "Plant_2_Generation_Data.csv"
)

plant2_weather = pd.read_csv(
    DATA_DIR / "Plant_2_Weather_Sensor_Data.csv"
)
```

```python
def merge_plant(generation, weather, plant_name):

    # Salin data agar data asli tidak berubah
    gen = generation.copy()
    wea = weather.copy()

    # Ubah nama kolom identitas
    gen = gen.rename(columns={
        "SOURCE_KEY": "INVERTER_KEY"
    })

    wea = wea.rename(columns={
        "SOURCE_KEY": "WEATHER_SENSOR_KEY"
    })

    # Gabungkan berdasarkan waktu dan ID pembangkit
    merged = gen.merge(
        wea,
        on=["DATE_TIME", "PLANT_ID"],
        how="left",
        validate="many_to_one",
        indicator=True
    )

    # Hitung hasil pencocokan
    print(f"\n{plant_name}")
    print("Total baris Generation:", len(gen))
    print(
        "Berhasil dicocokkan:",
        (merged["_merge"] == "both").sum()
    )
    print(
        "Tidak memiliki data cuaca:",
        (merged["_merge"] == "left_only").sum()
    )

    # Tambahkan identitas pembangkit
    merged["PLANT_NAME"] = plant_name

    # Hapus kolom indikator
    merged = merged.drop(columns="_merge")

    return merged
```

```python
plant1_merged = merge_plant(
    plant1_gen,
    plant1_weather,
    "Plant 1"
)

plant2_merged = merge_plant(
    plant2_gen,
    plant2_weather,
    "Plant 2"
)
```

```python
solar_combined = pd.concat(
    [plant1_merged, plant2_merged],
    ignore_index=True
)

print("Ukuran dataset gabungan:", solar_combined.shape)

display(solar_combined.head())
```

```python
# Membuat salinan dataset hasil merge
solar_cleaned = solar_combined.copy()

# Memastikan kolom tanggal bertipe datetime
solar_cleaned["DATE_TIME"] = pd.to_datetime(
    solar_cleaned["DATE_TIME"],
    errors="coerce"
)

# Memeriksa kualitas dataset
print("Jumlah baris:", len(solar_cleaned))
print("Jumlah kolom:", len(solar_cleaned.columns))
print("Jumlah duplikat:", solar_cleaned.duplicated().sum())

print("\nMissing values:")
display(solar_cleaned.isna().sum())

# Fitur waktu
solar_cleaned["DATE"] = solar_cleaned["DATE_TIME"].dt.date
solar_cleaned["YEAR"] = solar_cleaned["DATE_TIME"].dt.year
solar_cleaned["MONTH"] = solar_cleaned["DATE_TIME"].dt.month
solar_cleaned["HOUR"] = solar_cleaned["DATE_TIME"].dt.hour
solar_cleaned["DAY_NAME"] = solar_cleaned["DATE_TIME"].dt.day_name()

# Selisih suhu panel dan lingkungan
solar_cleaned["TEMP_DIFFERENCE"] = (
    solar_cleaned["MODULE_TEMPERATURE"] -
    solar_cleaned["AMBIENT_TEMPERATURE"]
)

# Selisih daya DC dan AC
solar_cleaned["POWER_DIFFERENCE"] = (
    solar_cleaned["DC_POWER"] -
    solar_cleaned["AC_POWER"]
)

# Rasio konversi DC ke AC
solar_cleaned["CONVERSION_RATIO"] = np.where(
    solar_cleaned["DC_POWER"] > 0,
    solar_cleaned["AC_POWER"] / solar_cleaned["DC_POWER"],
    np.nan
)

display(solar_cleaned.head())
```

```python
# Periksa jarak waktu antarobservasi setiap inverter

interval_check = (
    solar_cleaned
    .sort_values(["PLANT_NAME", "INVERTER_KEY", "DATE_TIME"])
    .groupby(["PLANT_NAME", "INVERTER_KEY"])["DATE_TIME"]
    .diff()
    .dt.total_seconds()
    .div(60)
)

print("Distribusi interval dalam menit:")
display(interval_check.value_counts().head(10))

print("\nJumlah interval bukan 15 menit:")
print((interval_check.dropna() != 15).sum())

# Jumlah timestamp unik setiap hari pada masing-masing pembangkit

timestamp_check = solar_cleaned.groupby(
    ["DATE", "PLANT_NAME"]
).agg(
    UNIQUE_TIMESTAMPS=("DATE_TIME", "nunique"),
    TOTAL_ROWS=("DATE_TIME", "count")
).reset_index()

display(timestamp_check.head(10))

print("Jumlah timestamp harian minimum:",
      timestamp_check["UNIQUE_TIMESTAMPS"].min())

print("Jumlah timestamp harian maksimum:",
      timestamp_check["UNIQUE_TIMESTAMPS"].max())

# Salin data dan urutkan berdasarkan inverter serta waktu
energy_data = solar_cleaned.copy()

energy_data = energy_data.sort_values(
    ["PLANT_NAME", "INVERTER_KEY", "DATE_TIME"]
)

# Ambil daya dan waktu observasi berikutnya
grouped = energy_data.groupby(
    ["PLANT_NAME", "INVERTER_KEY"]
)

energy_data["NEXT_TIME"] = grouped["DATE_TIME"].shift(-1)
energy_data["NEXT_AC_POWER"] = grouped["AC_POWER"].shift(-1)

# Hitung jarak waktu antarobservasi
energy_data["INTERVAL_MINUTES"] = (
    energy_data["NEXT_TIME"] - energy_data["DATE_TIME"]
).dt.total_seconds() / 60

# Pertahankan hanya interval tepat 15 menit
valid_interval = energy_data[
    energy_data["INTERVAL_MINUTES"] == 15
].copy()

# Estimasi energi menggunakan metode trapesium
valid_interval["ENERGY_KWH"] = (
    (valid_interval["AC_POWER"] +
     valid_interval["NEXT_AC_POWER"]) / 2
) * 0.25

# Ringkasan energi harian
daily_energy_valid = valid_interval.groupby(
    ["DATE", "PLANT_NAME"]
).agg(
    TOTAL_ENERGY_KWH=("ENERGY_KWH", "sum"),
    VALID_INTERVALS=("ENERGY_KWH", "count")
).reset_index()

display(daily_energy_valid.head(10))

# Ringkasan estimasi energi sebelumnya
old_energy = solar_cleaned.groupby(
    ["DATE", "PLANT_NAME"]
)["ENERGY_KWH"].sum().reset_index()

old_energy = old_energy.rename(
    columns={"ENERGY_KWH": "ENERGY_OLD_KWH"}
)

# Gabungkan dengan hasil perhitungan baru
energy_comparison = old_energy.merge(
    daily_energy_valid,
    on=["DATE", "PLANT_NAME"],
    how="left"
)

energy_comparison["ENERGY_DIFFERENCE"] = (
    energy_comparison["ENERGY_OLD_KWH"] -
    energy_comparison["TOTAL_ENERGY_KWH"]
)

display(energy_comparison.head(10))
```

```python
# Ringkasan total energi selama periode pengamatan

final_energy_summary = daily_energy_valid.groupby(
    "PLANT_NAME"
).agg(
    TOTAL_ENERGY_KWH=("TOTAL_ENERGY_KWH", "sum"),
    AVG_DAILY_ENERGY_KWH=("TOTAL_ENERGY_KWH", "mean"),
    MAX_DAILY_ENERGY_KWH=("TOTAL_ENERGY_KWH", "max"),
    MIN_DAILY_ENERGY_KWH=("TOTAL_ENERGY_KWH", "min"),
    TOTAL_VALID_INTERVALS=("VALID_INTERVALS", "sum")
).reset_index()

display(final_energy_summary.round(2))

# Setiap inverter memiliki maksimal 96 interval per hari.
# Jumlah inverter pada setiap pembangkit adalah 22.

expected_intervals = (
    solar_cleaned["DATE"].nunique() * 22 * 96
)

coverage_summary = daily_energy_valid.groupby(
    "PLANT_NAME"
)["VALID_INTERVALS"].sum().reset_index()

coverage_summary["EXPECTED_INTERVALS"] = expected_intervals

coverage_summary["COVERAGE_PERCENT"] = (
    coverage_summary["VALID_INTERVALS"] /
    coverage_summary["EXPECTED_INTERVALS"] * 100
)

display(coverage_summary.round(2))
```

```python
# Menghitung interval yang diharapkan berdasarkan
# timestamp pertama dan terakhir setiap inverter

expected_by_inverter = (
    solar_cleaned
    .groupby(["PLANT_NAME", "INVERTER_KEY"])
    .agg(
        FIRST_TIME=("DATE_TIME", "min"),
        LAST_TIME=("DATE_TIME", "max")
    )
    .reset_index()
)

expected_by_inverter["EXPECTED_INTERVALS"] = (
    (
        expected_by_inverter["LAST_TIME"]
        - expected_by_inverter["FIRST_TIME"]
    ).dt.total_seconds() / 900
).round().astype(int)

# Jumlah interval valid setiap inverter
valid_by_inverter = (
    valid_interval
    .groupby(["PLANT_NAME", "INVERTER_KEY"])
    .size()
    .reset_index(name="VALID_INTERVALS")
)

# Gabungkan kedua hasil
coverage_check = expected_by_inverter.merge(
    valid_by_inverter,
    on=["PLANT_NAME", "INVERTER_KEY"],
    how="left"
)

coverage_check["VALID_INTERVALS"] = (
    coverage_check["VALID_INTERVALS"].fillna(0)
)

# Ringkasan per pembangkit
coverage_final = coverage_check.groupby(
    "PLANT_NAME"
).agg(
    EXPECTED_INTERVALS=("EXPECTED_INTERVALS", "sum"),
    VALID_INTERVALS=("VALID_INTERVALS", "sum")
).reset_index()

coverage_final["COVERAGE_PERCENT"] = (
    coverage_final["VALID_INTERVALS"]
    / coverage_final["EXPECTED_INTERVALS"] * 100
)

display(coverage_final.round(2))
```
